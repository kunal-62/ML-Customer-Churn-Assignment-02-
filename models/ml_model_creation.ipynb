## Data Pre-processing

```python
data.info()
```

```python
# Convert 'TotalCharges' to numeric, coercing errors to NaN
data['TotalCharges'] = pd.to_numeric(data['TotalCharges'], errors='coerce')

# Check for missing values after conversion
missing_values = data.isnull().sum()
print("Missing values after converting TotalCharges:")
print(missing_values[missing_values > 0])
```

```python
# Drop rows where 'TotalCharges' is NaN
data.dropna(inplace=True)

# Verify the DataFrame's shape after dropping rows
print(f"DataFrame shape after dropping rows with missing TotalCharges: {data.shape}")
```

### Encoding Categorical Features

Many of the features are categorical and need to be converted into numerical format for machine learning models. We'll use one-hot encoding for multi-category features and label encoding for binary features where appropriate.

```python
# Identify categorical columns (object type, excluding 'customerID')
categorical_cols = data.select_dtypes(include='object').columns.tolist()
categorical_cols.remove('customerID') # customerID is an identifier, not a feature

print("Categorical columns and their unique values:")
for col in categorical_cols:
    print(f"\n{col}: {data[col].unique()}")
```

```python
# Standardize 'No internet service' and 'No phone service' to 'No'


for col in ['OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies']:
    data[col] = data[col].replace({'No internet service': 'No'})
data['MultipleLines'] = data['MultipleLines'].replace({'No phone service': 'No'})

# Map binary categorical columns to 0 and 1
# 'gender' mapping: 'Female' -> 0, 'Male' -> 1
data['gender'] = data['gender'].map({'Female': 0, 'Male': 1})

# Other binary columns mapping: 'No' -> 0, 'Yes' -> 1
binary_cols = ['Partner', 'Dependents', 'PhoneService', 'PaperlessBilling', 'Churn']
for col in binary_cols:
    data[col] = data[col].map({'No': 0, 'Yes': 1})

print("DataFrame head after standardizing and binary encoding:")
print(data.head())
```

```python
# One-hot encode remaining categorical columns
# Exclude customerID, already encoded binary columns, and the target 'Churn'
columns_to_onehot = [col for col in data.select_dtypes(include='object').columns if col not in ['customerID']]

data = pd.get_dummies(data, columns=columns_to_onehot, drop_first=True)

# Drop customerID column
data.drop('customerID', axis=1, inplace=True)

print("\nDataFrame head after one-hot encoding and dropping customerID:")
print(data.head())
print("\nDataFrame info after all preprocessing:")
```

### Splitting Dataset into Training and Testing dataset  

```python
# 2. Separate features (X) from the target label (y)
# Replace 'churn' with the exact name of your target column

X = data.iloc[:, :-1]
y = data.iloc[:, -1]

# 3. Automatically split into Train (80%) and Test (20%) datasets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.20,      # 20% goes to the testing set, 80% goes to training
    random_state=42,     # Locks the randomness so you get the exact same split every time you run the code
    stratify=y           # Highly recommended: keeps the same % of Yes/No churn in both train and test
)

print(f"Training features shape: {X_train.shape}")
print(f"Testing features shape: {X_test.shape}")

# ---------------- Saving a test.csv file ----------------

# 1. Combine test features and target labels back into one DataFrame
test_dataset = X_test.copy()
test_dataset['churn'] = y_test  # Replace 'churn' with your actual target column name

proj_root_dir = str(os.getcwd()).replace('models','')
file_path = os.path.join(proj_root_dir, "test_data.csv")

# 2. Save it directly into your project folder
test_dataset.to_csv(file_path, index=False)

print(f"test_data.csv has been successfully saved to path {proj_root_dir}!")
```

### Scaling Trained Dataset

```python
# Apply Standard Scaling to features
import joblib

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)

# Save the fitted scaler
joblib.dump(scaler, "scaler.joblib")

# Convert scaled arrays back to DataFrames with column names for better readability
X_train_scaled = pd.DataFrame(X_train_scaled, columns=X_train.columns)

print("\nFirst 5 rows of scaled X_train:")
print(X_train_scaled.head())
```

## 1. Logistic Regression

```python
from sklearn.linear_model import LogisticRegression

# Initialize and train the Logistic Regression model
log_reg_model = LogisticRegression(random_state=42)
log_reg_model.fit(X_train_scaled, y_train)

# Save your trained model to a file
joblib.dump(log_reg_model, "log_reg_model.joblib")
```

## 2. Decision Tree Classifier

```python
from sklearn.tree import DecisionTreeClassifier

# Initialize and train the Decision Tree Classifier
dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(X_train_scaled, y_train)

# Save your trained model to a file
joblib.dump(dt_model, "dt_model.joblib")
```

## 3. k-Nearest Neighbors (kNN)

```python
from sklearn.neighbors import KNeighborsClassifier

# Initialize and train the kNN model
# Using n_neighbors=5 as a common starting point
knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train_scaled, y_train)

# Save your trained model to a file
joblib.dump(knn_model, "kNN_model.joblib")
```

## 4. Naive Bayes

```python
from sklearn.naive_bayes import GaussianNB

# Initialize and train the Gaussian Naive Bayes model
nb_model = GaussianNB()
nb_model.fit(X_train_scaled, y_train)

# Save your trained model to a file
joblib.dump(nb_model, "nb_model.joblib")
```

## 5. Random Forest(Ensemble)

```python
from sklearn.ensemble import RandomForestClassifier

# Initialize and train the Random Forest Classifier
rf_model = RandomForestClassifier(random_state=42)
rf_model.fit(X_train_scaled, y_train)

# Save your trained model to a file
joblib.dump(rf_model, "rf_model.joblib")
```

## 6. Support Vector Machine

```python
from sklearn.svm import SVC

# Initialize and train the Support Vector Machine model
# Using probability=True to enable AUC calculation
svm_model = SVC(random_state=42, probability=True)
svm_model.fit(X_train_scaled, y_train)

# Save your trained model to a file
joblib.dump(svm_model, "svm_model.joblib")
```
