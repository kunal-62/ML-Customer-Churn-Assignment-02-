### kNN Model

```python
import numpy as np
import pandas as pd
import joblib

from sklearn.model_selection import GridSearchCV
from sklearn.neighbors import KNeighborsClassifier
```

```python
X_train = pd.read_csv("G:\Telco_Customer_Churn_ML_Asgn02\models\saved_data\X_train.csv")
y_train = pd.read_csv("G:\Telco_Customer_Churn_ML_Asgn02\models\saved_data\y_train.csv").squeeze()
```

```python
scaler = joblib.load("scaler.joblib")

X_train_scaled = scaler.fit_transform(X_train)

# Convert scaled arrays back to DataFrames with column names for better readability
X_train_scaled = pd.DataFrame(X_train_scaled, columns=X_train.columns)

print("\nFirst 5 rows of scaled X_train:")
print(X_train_scaled.head())
```

#### Tune kNN

```python
# Initialize kNN
knn_model = KNeighborsClassifier()

# Small but useful hyperparameter grid
param_grid_knn = {
    "n_neighbors": [3, 5, 7, 9, 11, 15],
    "weights": ["uniform", "distance"],
    "metric": ["euclidean", "manhattan"]
}

# Grid Search
grid_search_knn = GridSearchCV(
    estimator=knn_model,
    param_grid=param_grid_knn,
    scoring="f1",
    cv=5,
    n_jobs=-1,
    verbose=1
)

# Train
grid_search_knn.fit(X_train_scaled, y_train)

# Results
print("Best parameters:")
print(grid_search_knn.best_params_)

print("\nBest CV F1:")
print(grid_search_knn.best_score_)
```

```python
best_knn_model = grid_search_knn.best_estimator_

print("Best parameters:")
print(grid_search_knn.best_params_)

print("Best CV F1:")
print(grid_search_knn.best_score_)

joblib.dump(
    best_knn_model,
    "kNN_model.joblib"
)
```
