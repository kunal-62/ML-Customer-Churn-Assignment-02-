### Naive Bayes Model

```python
import numpy as np
import pandas as pd
import joblib

from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import GridSearchCV
```

```python
X_train = pd.read_csv("G:\Telco_Customer_Churn_ML_Asgn02\models\saved_data\X_train.csv")
y_train = pd.read_csv("G:\Telco_Customer_Churn_ML_Asgn02\models\saved_data\y_train.csv").squeeze()
```

```python
scaler = joblib.load("scaler.joblib")

X_train_scaled = scaler.fit_transform(X_train)

# Convert scaled arrays back to DataFrames with column names for better readability
X_train_scaled = pd.DataFrame(X_train_scaled, columns=X_train.columns)

print("\nFirst 5 rows of scaled X_train:")
print(X_train_scaled.head())
```

#### Tune Naive Bayes Classifier

```python
# Initialize Gaussian Naive Bayes
nb_model = GaussianNB()

# Small hyperparameter grid
param_grid_nb = {
    "var_smoothing": [1e-11, 1e-10, 1e-9, 1e-8, 1e-7]
}

# Grid Search
grid_search_nb = GridSearchCV(
    estimator=nb_model,
    param_grid=param_grid_nb,
    scoring="f1",
    cv=5,
    n_jobs=-1,
    verbose=1
)

# Train
grid_search_nb.fit(X_train_scaled, y_train)

# Results
print("Best parameters:")
print(grid_search_nb.best_params_)

print("\nBest CV F1:")
print(grid_search_nb.best_score_)
```

### Saving the Naive Bayes Classifier

```python
best_nb_model = grid_search_nb.best_estimator_

joblib.dump(
    best_nb_model,
    "nb_model_tuned.joblib"
)
```
