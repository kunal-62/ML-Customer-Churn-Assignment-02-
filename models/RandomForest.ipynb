### Random Forest Classifier 

```python
import numpy as np
import pandas as pd
import joblib

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV
```

```python
X_train = pd.read_csv("G:\Telco_Customer_Churn_ML_Asgn02\models\saved_data\X_train.csv")
y_train = pd.read_csv("G:\Telco_Customer_Churn_ML_Asgn02\models\saved_data\y_train.csv").squeeze()
```

#### Note: Random Forest doesn't need scaling. Just like your Decision Tree, Random Forest is tree-based.

```python
# Initialize Random Forest
rf_model = RandomForestClassifier(
    random_state=42,
    n_jobs=-1
)

# Small but useful hyperparameter grid
param_grid_rf = {
    "n_estimators": [100, 200],
    "max_depth": [5, 10, None],
    "min_samples_leaf": [1, 5],
    "max_features": ["sqrt", "log2"],
    "class_weight": [None, "balanced"]
}

# Grid Search
grid_search_rf = GridSearchCV(
    estimator=rf_model,
    param_grid=param_grid_rf,
    scoring="f1",
    cv=5,
    n_jobs=-1,
    verbose=1
)

# Train
grid_search_rf.fit(X_train, y_train)

print("Best parameters:")
print(grid_search_rf.best_params_)

print("\nBest CV F1:")
print(grid_search_rf.best_score_)
```

```python
best_rf_model = grid_search_rf.best_estimator_

joblib.dump(
    best_rf_model,
    "rf_model.joblib"
)
```
